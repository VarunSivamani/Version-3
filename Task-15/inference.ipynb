```python
import torch
from transformers import AutoTokenizer
from deepseek_smollm2 import DeepSeekModel, DeepSeekConfig
from collections import OrderedDict

# Device configuration
device = "cuda:1" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("HuggingFaceTB/cosmo2-tokenizer")

# Load model
model = DeepSeekModel(DeepSeekConfig())

# Load checkpoint and fix key mismatches caused by torch.compile
checkpoint = torch.load("model.pt", map_location=device)
new_state_dict = OrderedDict()
for k, v in checkpoint.items():
    new_key = k.replace("_orig_mod.", "")  # Remove the prefix
    new_state_dict[new_key] = v

model.load_state_dict(new_state_dict)
model.to(device)
model.eval()

# Define input text for inference
input_text = "This is an inference example."
encoded_text = tokenizer(input_text, return_tensors="pt").to(device)

# Generate predictions
with torch.no_grad():
    max_new_tokens = 50
    temperature = 1.0
    top_k = 40
    
    prediction_logits = model.generate(encoded_text["input_ids"], max_new_tokens=max_new_tokens, temperature=temperature, top_k=top_k)
    prediction = tokenizer.decode(prediction_logits[0], skip_special_tokens=True)
    
    print(f"Generated Output:\n{prediction}")
```
